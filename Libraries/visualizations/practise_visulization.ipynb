```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[2,4,6,8,10,12]
z=[0,3,6,9,12,15]
plt.plot(x, y)

plt.title("Axes1")
plt.xlabel("X ")
plt.ylabel("Y ")

plt.show()
```

```python
fig, ax = plt.subplots(1,2,figsize=(10,5))

ax[0].plot(x,y)
ax[1].plot(x,z)
```

```python
#plt.scatter(x, y)
plt.scatter(x, y)
plt.title("scatter")
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.figure(figsize=(6,4))
plt.plot(x, y, marker='o', color='blue')

plt.title("Example Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.plot(x, y, marker="o",color="red")

plt.show()
```
