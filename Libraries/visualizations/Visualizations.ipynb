## Matplotlib – Core Foundations

#### What is Matplotlib?

**Matplotlib** is the fundamental Python library for creating static visualizations, giving full control over every element in the plot.

***Think of it as:** “The low-level engine behind almost every Python visualization.”*

#### When to Use Matplotlib

Use Matplotlib when:

- You need full customization

- You want multi-plot layouts

- You need precise control

- You want publication-quality figures

#### Strengths

- Extremely flexible

- Works with any data structure

- Many plot types

- Limitations

- More code

- Styling requires manual effort

#### Limitations

- More code

- Styling requires manual effort

#### Matplotlib Structure

- Figure → the canvas

- Axes → the actual plot

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1, 2, 3], [3, 2, 1])
axes[0].set_title("First Axes")

axes[1].plot([1, 2, 3], [1, 4, 9])
axes[1].set_title("Second Axes")

plt.show()

```

## Core Matplotlib Plots 

For each plot, we follow the same steps:

- Prepare data

- Call the plotting function

- Label & customize

- Interpret the insight

#### 1️⃣ Line Plot

**What it shows:** Trend over time / continuous variable

**Example use case:** Monthly sales trend

**How to build it:**

- X-axis → time (months)
    
- Y-axis → numeric value (sales)
    
- Use markers to highlight points
    

**Code example:**

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 220, 250, 270, 300]

plt.plot(months, sales, marker='o', linestyle='-', color='blue')
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.grid(True)
plt.show()
```

**Insight to mention:** Is the trend increasing, decreasing, or stable?

#### 2️⃣ Scatter Plot

**What it shows:** Relationship between two numeric variables

**Example use case:** Study hours vs exam score

**How to build it:**

- X-axis → first numeric variable
    
- Y-axis → second numeric variable
    
- Adjust point size to improve readability
    

**Code example:**

```python
hours = [1, 2, 3, 4, 5, 6]
scores = [55, 60, 65, 70, 75, 80]

plt.scatter(hours, scores, s=100, color='green')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

**Insight to mention:** Is there correlation? Any outliers?

#### 3️⃣ Bar Plot

**What it shows:** Comparison across categories

**Example use case:** Number of students per major

**How to build it:**

- X-axis → categories
    
- Y-axis → counts or values
    
- Add labels on bars if needed
    

**Code example:**

```python
majors = ["CS", "Math", "Physics", "Biology"]
students = [50, 30, 20, 40]

plt.bar(majors, students, color='orange')
plt.xlabel("Major")
plt.ylabel("Number of Students")
plt.title("Students per Major")
plt.show()
```

**Insight to mention:** Which category is highest or lowest?

#### 4️⃣ Histogram

**What it shows:** Distribution of numeric data

**Example use case:** Age distribution of students

**How to build it:**

- One numeric column
    
- Choose appropriate number of bins
    

**Code example:**

```python
ages = [18, 19, 18, 20, 21, 22, 19, 18, 20, 21]

plt.hist(ages, bins=5, edgecolor='black', color='skyblue')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

**Insight to mention:** Is the data skewed? Where is the concentration?

#### 5️⃣ Box Plot

**What it shows:** Median, quartiles, and outliers

**Example use case:** Exam scores distribution

**How to build it:**

- Pass numeric data
    
- Optionally compare multiple groups
    

**Code example:**

```python
scores = [55, 60, 65, 70, 72, 75, 78, 90, 120]

plt.boxplot(scores)
plt.title("Exam Scores Boxplot")
plt.show()
```


**Insight to mention:** Are there outliers? How spread out is the data?

#### 6️⃣ Subplots

**What it shows:** Multiple plots in one figure

**Example use case:** Comparing sales and profit separately

**How to build it:**

- Define rows and columns
    
- Plot each chart on a separate Axes
    

**Code example:**


```python
import numpy as np

x = np.arange(1, 6)
sales = [200, 220, 250, 270, 300]
profit = [40, 45, 60, 65, 80]

fig, axes = plt.subplots(2, 1, figsize=(8, 6))
axes[0].plot(x, sales)
axes[0].set_title("Sales")
axes[1].plot(x, profit)
axes[1].set_title("Profit")
plt.tight_layout()

plt.show()
```

**Insight to mention:** How do trends compare between plots?

#### 7️⃣ Bar + Line Combination

**What it shows:** Quantity vs trend together

**Example use case:** Sales (bars) vs profit (line)

**How to build it:**

- Use same Axes
    
- Draw bars first, then line
    
- Add legend
    

**Code example:**

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 300, 350]
profit = [50, 70, 90, 120]

fig, ax = plt.subplots()
ax.bar(months, sales, color='lightblue', label='Sales')
ax.plot(months, profit, color='red', marker='o', label='Profit')
ax.set_title("Sales vs Profit")
ax.legend()

plt.show()
```

**Insight to mention:** Does profit follow the same trend as sales?

#### Pie Chart

**What it shows:** Proportions / percentage of a whole

**Example use case:** Distribution of students across departments

**How to build it:**

- Prepare categories (labels)

- Prepare values

- Use plt.pie()

- Add percentages for clarity

**Code example:**

```python
import matplotlib.pyplot as plt

departments = ['CS', 'IT', 'IS', 'AI']
students = [40, 25, 20, 15]

plt.figure(figsize=(5, 5))
plt.pie(
    students,
    labels=departments,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Students Distribution by Department")
plt.show()
```

> ⚠️ **Note on Pie Charts**  
> Pie charts exist, but are generally avoided in analytical reporting because humans compare lengths better than angles.

##  Transition: Why Move to Seaborn?

**After using Matplotlib, we notice:**

- A lot of repetitive code
    
- Manual styling effort
    

That’s why **Seaborn** exists.

Key differences from Matplotlib:

| Feature                   | Matplotlib                   | Seaborn                                          |
| ------------------------- | ---------------------------- | ------------------------------------------------ |
| API Level                 | Low-level (detailed control) | High-level (less code, more style)               |
| Default Styles / Palettes | Basic / need customization   | Beautiful default themes & palettes              |
| Statistical Plots         | Limited / need manual work   | Built-in: regression, distribution, violin, etc. |
| Ease of Use               | More code for customization  | Less code, more readable                         |


**When to move from Matplotlib to Seaborn:**

- Quick exploration of data (EDA)

- Visualizations with statistical summaries

- Beautiful plots with minimal customization

- Ready-made themes and color palettes

#### 1️⃣ Countplot

**What it shows:** Counts of each category automatically

**Example use case:** Number of bills per day

**How to build it:**

- Choose categorical column for x
    
- Optionally use hue for comparison
    

**Code example:**

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.countplot(x="day", data=tips, palette="pastel", hue = "day")
plt.title("Number of Bills per Day")
plt.show()
```

**Explanation:**

countplot → Counts occurrences of each category automatically.

palette → Colors of the bars.

Perfect for categorical data exploration.

**Insight to mention:** Which day has the highest number of customers?


#### 2️⃣ Barplot (Aggregated)

**What it shows:** Mean (or other estimator) per category

**Example use case:** Average total bill per day

**How to build it:**

- x → category
    
- y → numeric column
    
- estimator → mean (default) or median
    

**Code example:**

```python
sns.barplot(x="day", y="total_bill", data=tips, palette="coolwarm" , hue = "day")
plt.title("Average Total Bill per Day")
plt.show()
```

**Explanation:**

y → Numeric value, automatically aggregated (default: mean).

Shows confidence interval by default.

#### 3️⃣ Histogram / Histplot

**What it shows:** Distribution of numeric variable

**Example use case:** Distribution of total bill

**How to build it:**

- Choose numeric column
    
- Control bins and KDE
    

**Code example:**

```python
sns.histplot(tips["total_bill"], bins=20, kde=True, color="skyblue")
plt.title("Distribution of Total Bill")
plt.show()
```

**Explanation:**

kde=True → Adds smooth density curve.

bins → Number of intervals.

**Insight to mention:** Is the distribution skewed or symmetric?

#### 4️⃣ KDE Plot

**What it shows:** Smooth density estimation of distribution

**Example use case:** Shape of total bill distribution

**How to build it:**

- Use numeric column
    
- fill=True for better visualization
    

**Code example:**

```python
sns.kdeplot(tips["total_bill"], fill=True, color="blue")
plt.title("KDE of Total Bill")
plt.show()
```

**Insight to mention:** Where is the highest density of values?

#### 5️⃣ Boxplot

**What it shows:** Median, quartiles, and outliers per category

**Example use case:** Total bill by day

**How to build it:**

- x → category
    
- y → numeric column
    

**Code example:**



```python
sns.boxplot(x="day", y="total_bill", data=tips, palette="Set2", hue = "day")
plt.title("Total Bill Distribution by Day")
plt.show()
```

**Explanation:**

Shows median, quartiles, outliers per category.

**Insight to mention:** Which day has more outliers or higher spread?

#### 6️⃣ Violin Plot

**What it shows:** Distribution + density + boxplot

**Example use case:** Detailed distribution of bills per day

**How to build it:**

- Similar to boxplot
    
- Better for distribution shape
    

**Code example:**


```python
sns.violinplot(x="day", y="total_bill", data=tips, palette="muted" , hue = "day")
plt.title("Violin Plot of Total Bill by Day")
plt.show()
```

**Explanation:**

More detailed distribution than boxplot.

Useful to see shape of data distribution per category.

**Insight to mention:** How does the shape of distribution differ across days?

#### 7️⃣ Pairplot

**What it shows:** Relationships between all numeric variables

**Example use case:** Overall dataset overview

**How to build it:**

- Pass full DataFrame
    
- Use hue for categorical split
    

**Code example:**


```python
sns.pairplot(tips, hue="sex", palette="bright")
plt.show()
```

**Explanation:**

Shows scatter plots for each pair of numeric columns.

hue → Color by category.

**Insight to mention:** Which variables seem strongly related?

#### 8️⃣ Jointplot

**What it shows:** Relationship between two variables + distributions

**Example use case:** Total bill vs tip

**How to build it:**

- Specify x and y
    
- Choose kind (scatter, kde, hex)
    

**Code example:**


```python
sns.jointplot(x="total_bill", y="tip", data=tips, kind="scatter", color="purple")
plt.show()
```

**Explanation:**

kind="scatter" → Scatter plot (can also use "kde" or "hex").

Good for analyzing relationships between two variables with distribution context.

**Insight to mention:** Does tip increase as total bill increases?

#### 9️⃣ Heatmap

**What it shows:** Correlation matrix

**Example use case:** Relationship between numeric variables

**How to build it:**

- Compute correlation
    
- Use annot to show values
    

**Code example:**

```python
corr = tips.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

**Explanation:**

annot=True → Shows values in cells.

cmap → Colormap.

Very useful to see strength of relationships.

---
**Insight to mention:** Which variables have strong positive or negative correlation?

## Advanced Visuals 

### FacetGrid

- Same plot split by category
    

### lmplot

- Regression line by category
    

### Swarm vs Strip Plot

- Strip: simple, overlapping points
    
- Swarm: non-overlapping, clearer distribution
    

> Swarm plot = smarter strip plot

##### 1) FacetGrid — Multiple Plots by Category

**What is FacetGrid?**

FacetGrid allows you to split your data into multiple subplots automatically based on a categorical feature.
Useful when you want to compare the same plot across groups (e.g., male vs female).


```python
sns.FacetGrid(tips, col="sex").map(sns.histplot, "total_bill")
plt.show()
```

##### 2) lmplot — Regression Plot by Category

**What is lmplot?**

A line of best fit + scatter plot.
Great for showing relationships AND trends.

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex")
plt.title("Bill vs Tip with Trendline")
plt.show()
```

##### 3) Swarm Plot vs Strip Plot

**Strip Plot**

- Points may overlap

- Simple

- Shows distribution of points

**Swarm Plot**

- Spreads points so they never overlap

- Better readability

- More accurate representation

```python
plt.figure(figsize=(6,4))
sns.swarmplot(x="day", y="total_bill", data=tips)
plt.title("Swarm Plot")
plt.show()
```

“Swarm plot is like a smarter version of strip plot — it avoids point overlap, so we see the distribution more clearly.”

```python
plt.figure(figsize=(6,4))
sns.stripplot(x="day", y="total_bill", data=tips)
plt.title("Strip Plot")
plt.show()
```
